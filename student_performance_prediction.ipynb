```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

#Loading data set
df = pd.read_csv("Final_Marks_Data.csv")

#2.Categorize Final Exam Marks into High, Average, and Low
def categorize_performance(marks):
  if marks>= 75:
    return 'High'
  elif marks >= 55:
    return 'Average'
  else:
    return 'Low'

df['Performance_Category'] = df['Final Exam Marks (out of 100)'].apply(categorize_performance)

#3. Select features (x) and target (y)
x = df[['Attendance (%)', 'Internal Test 1 (out of 40)', 'Internal Test 2 (out of 40)', 'Assignment Score (out of 10)', 'Daily Study Hours']]
y = df['Performance_Category']

#4. Split into training and testing sets(80% train, 20% test)
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

#5.Train the Random Forest Model
model = RandomForestClassifier(random_state=42)
model.fit(x_train, y_train)

#6.Test and Evaluate
y_pred = model.predict(x_test)
print("Model Accuracy:", f"{accuracy_score(y_test, y_pred):.2%}")
print("\nClassification Report:\n", classification_report(y_test, y_pred))


```

```python
# A PREDICTION BLOCK TO TEST A BRAND NEW STUDENT SCORES
#Test a brand new students scores

sample_student = [[85, 35, 36, 9, 4.5]]

predicted_category = model.predict(sample_student)
print("Predicted Performance Category:", predicted_category[0])
```
